```python
## helper functions
def create_answer(
        level,
        label,
        score,
        seal):

    return {
        "level": level,
        "label": label,
        "score": score,
        "seal": seal
    }

def create_criterion(
        cid,
        title,
        description,
        answers):

    return {
        "id": cid,
        "title": title,
        "description": description,
        "allow_na": True,
        "binding": False,
        "answers": answers
    }
```
